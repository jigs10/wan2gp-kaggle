# Wan2GP on Kaggle

Runs Wan2GP in a Kaggle notebook and saves all model downloads to `/kaggle/tmp` instead of the working directory.

**Run the cells in order, top to bottom.**

## 1. Download Wan2GP

Clones the Wan2GP repo into `/kaggle/working`. If it's already there, it pulls the latest changes instead.

```python
%cd /kaggle/working
import os
if not os.path.exists('/kaggle/working/Wan2GP'):
    !git clone https://github.com/deepbeepmeep/Wan2GP.git
else:
    !git -C /kaggle/working/Wan2GP pull
%cd /kaggle/working/Wan2GP
```

## 2. Install requirements

Installs all the Python packages Wan2GP needs.

```python
!pip install -r requirements.txt
```

## 3. Set model download path to /kaggle/tmp

Wan2GP saves models in a `ckpts` folder. This cell links that folder to `/kaggle/tmp/ckpts`, so models go to the big temporary disk and don't fill the 20GB working directory. It also sends the Hugging Face cache to `/kaggle/tmp`.

Run this **before** launching the app.

```python
%cd /kaggle/working/Wan2GP
!mkdir -p /kaggle/tmp/ckpts /kaggle/tmp/hf
!mv ckpts/* /kaggle/tmp/ckpts/ 2>/dev/null; rm -rf ckpts; ln -sfn /kaggle/tmp/ckpts ckpts
%env HF_HOME=/kaggle/tmp/hf
!df -h /kaggle/tmp /kaggle/working
```

## 4. Launch Wan2GP

Starts the app and gives you a public Gradio link. Models download to `/kaggle/tmp` the first time you use them.

```python
%cd /kaggle/working/Wan2GP
!python wgp.py --share --profile 5
```
